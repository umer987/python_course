### 44. User ka weight aur height le kar BMI category batayein (Underweight, Normal, Overweight).

```python
# Write your code here

weight = float(input("Apna weight (kg mein) enter karein: "))
height_cm = float(input("Apni height (cm mein) enter karein: "))

# Height ko centimeters se meters mein convert karna
height_m = height_cm / 100

# BMI ka formula: weight / (height * height)
bmi = weight / (height_m ** 2)

print(f"Aapka BMI hai: {bmi:.2f}")

if bmi < 18.5:
    print("Category: Underweight")
elif bmi >= 18.5 and bmi < 25:
    print("Category: Normal")
elif bmi >= 25 and bmi < 30:
    print("Category: Overweight")
else:
    print("Category: Obese")
```

### 60. Final Challenge: Ek mini-project logic banayein jisme user se Username, Password, aur ek Security Question poochein nested loops aur if-else ke saath.
