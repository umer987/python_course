# **Q7:** `while loop` ka use karke user se tab tak numbers input lein jab tak wo `0` enter na karde. End mein saare numbers ka total dikhayein.

````python
# type your code here

total_sum = 0

print("Numbers enter karte jayein (Stop karne ke liye 0 enter karein):")

while True:
    number = float(input("Number enter karein: "))
    
    if number == 0:
        break  # Jab user 0 enter karega toh loop khatam ho jayega
        
    total_sum += number

print(f"\n[STOP] Aapne 0 enter kiya.")
print(f"Saare enter kiye gaye numbers ka total sum: {total_sum}")
````

# **Q8:** Ek list `numbers = [12, 45, 7, 23, 56, 90, 11]` mein se sabse bada (Maximum) number find karein loop ke zariye.

````python
# type your code here
numbers = [12, 45, 7, 23, 56, 90, 11]


max_num = numbers[0]

for num in numbers:
    if num > max_num:
        max_num = num

print(f"List mein se sabse bada (Maximum) number hai: {max_num}")
````

# **Q9:** User se ek string input lein (e.g., 'Python') aur loop ka use karke us string ko reverse print karein.

````python
# type your code here
string_input = input("Ek string enter karein (e.g., Python): ")

reversed_string = ""


for char in string_input:
    reversed_string = char + reversed_string

print(f"Reversed string: {reversed_string}")
````

# **Q10:** 1 se 50 tak ke numbers check karein. Agar number 3 se divide hota hai to 'Fizz' print karein, agar 5 se hota hai to 'Buzz' print karein, aur agar dono se hota hai to 'FizzBuzz' print karein.

# **Q11:** Ek list di gayi hai `items = [1, 2, 3, 4, 5]`. Loop ka use karke ek nayi list banayein jis mein har number ka **Square** (e.g., 1, 4, 9...) ho.

````python
# type your code here

items = [1, 2, 3, 4, 5]
squares = []

# Loop ke zariye har number ka square nikal kar nayi list mein append karenge
for num in items:
    squares.append(num ** 2)

print(f"Original List: {items}")
print(f"Squares List: {squares}")
````

# **Q12:** User se ek number lein aur check karein ke wo **Prime Number** hai ya nahi (using loops).

````python
# type your code here
number = int(input("Ek number enter karein: "))


if number > 1:
    is_prime = True
    
    
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            is_prime = False
            break 
            
    if is_prime:
        print(f"{number} ek Prime Number hai.")
    else:
        print(f"{number} Prime Number nahi hai.")
else:
    print(f"{number} Prime Number nahi hai (Prime numbers hamesha 1 se bade hote hain).")
````

# **Q16:** User se ek number input lein aur us ka **Factorial** find karein (e.g., 5! = 5*4*3*2*1).

````python
# type your code here
number = int(input("Ek number enter karein: "))

factorial = 1

if number < 0:
    print("Negative numbers ka factorial nahi hota.")
elif number == 0:
    print("0 ka factorial 1 hota hai.")
else:
    for i in range(1, number + 1):
        factorial *= i
    print(f"{number} ka factorial {factorial} hai.")
````

# **Q18:** Ek manual counting karein ke ek string (e.g., 'Learning Python is fun') mein kitne **Vowels** (a, e, i, o, u) hain.

````python
# type your code here
sentence = "Learning Python is fun"
vowels = "aeiouAEIOU"
vowel_count = 0

for char in sentence:
    if char in vowels:
        vowel_count += 1

print(f"String mein total {vowel_count} vowels hain.")
````

# **Q19:** **Fibonacci Sequence** ke pehle 10 numbers print karein (0, 1, 1, 2, 3, 5, 8...)

````python
# type your code here
n = 10
a, b = 0, 1

for i in range(n):
    print(a, end=" " if i < n - 1 else "\n")
    a, b = b, a + b
````

# **Q20:** Ek 'Number Guessing Game' banayein jahan program ek number (1-20) fix kare aur user ko 3 chances de guess karne ke liye. `while loop` aur `break` ka use karein.

````python
target_number = 12
chances = 3
attempt = 1

print("--- Number Guessing Game ---")
print("Maine 1 se 20 ke beech ek number socha hai. Aapke paas 3 chances hain!")

while attempt <= chances:
    guess = int(input(f"Attempt {attempt}: Apna guess enter karein: "))
    
    if guess == target_number:
        print("Congratulations! Aapne bilkul sahi number guess kiya!")
        break
    elif guess < target_number:
        print("Aapka guess thoda chota (low) hai.")
    else:
        print("Aapka guess thoda bada (high) hai.")
        
    attempt += 1
else:
    print(f"\nGame Over! Aapki saari chances khatam ho gayin. Sahi number {target_number} tha.")
````
